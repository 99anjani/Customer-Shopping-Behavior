```python
df = df.drop_duplicates()
```

```python
df.shape
```

```python
df.isnull().sum() #Missing values
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_') #snake_case
df = df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})
```

```python
df.columns
```

```python
#create a column age_group
labels = ['Young Adult','Adult', 'Middle Aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)

df[['age', 'age_group']].head(10)
```

```python
df['frequency_of_purchases'].unique()
```

```python
#create column purchase_frequency_days

freq_mapping = {'Fortnightly': 14, 'Weekly' : 7, 'Annually': 365, 'Quarterly':90, 'Bi-Weekly':14,
       'Monthly':30, 'Every 3 Months':90}

df['purchases_frequency_days'] = df['frequency_of_purchases'].map(freq_mapping)
```

```python
df[['purchases_frequency_days','frequency_of_purchases']].head(10)
```

```python
# check discount_applied column and promo_code_used columns are same or not
(df['discount_applied']== df['promo_code_used']).all()
```

```python
df = df.drop('promo_code_used', axis=1)
```
